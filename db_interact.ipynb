# 1. Conectarse y ver la lista de tablas disponibles

```python
import sqlite3

# Reemplaza 'tu_archivo.db' con la ruta exacta de tu archivo
conexion = sqlite3.connect("data/hotel_data.db")
cursor = conexion.cursor()

# Consultar las tablas existentes en la base de datos
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
tablas = cursor.fetchall()

print("Tablas encontradas:")
for tabla in tablas:
    print("-", tabla[0])
```

# 2. Explorar las filas y columnas de una tabla

```python
import pandas as pd

# Cargar una tabla específica en un DataFrame de pandas para explorarla visualmente
nombre_tabla = "bar_by_roomtype"
df = pd.read_sql_query(f"SELECT * FROM {nombre_tabla}", conexion)

# Muestra las primeras filas de forma ordenada
display(df)
```

# 3. Borrar una tabla

Para eliminar una tabla por completo de forma definitiva, ejecutar el comando DROP TABLE:

```python
# Reemplaza 'tabla_a_borrar' por el nombre real de la tabla que deseas eliminar
tabla_a_borrar = "tabla_a_borrar"

cursor.execute(f"DROP TABLE IF EXISTS {tabla_a_borrar};")
conexion.commit()

print(f"La tabla '{tabla_a_borrar}' ha sido borrada exitosamente.")

# Opcional: cierra la conexión cuando termines
conexion.close()
```
